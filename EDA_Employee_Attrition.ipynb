```python
# Thiranex Data Science Internship
# Task 3: Exploratory Data Analysis (EDA)
# Project: Employee Attrition Analysis

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

print("Libraries imported successfully!")
```

```python
# Load the employee attrition dataset

df = pd.read_csv("../Thiranex_Task2_Predictive_Modeling/employee_attrition_data.csv")

print("Dataset loaded successfully!")
print("Dataset Shape:", df.shape)

df.head()
```

```python
# Data Understanding

print("Dataset Information:")
print(df.info())

print("\nStatistical Summary:")
display(df.describe())

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:")
print(df.duplicated().sum())
```

```python
# Data Preparation

# Check the unique values in categorical/binary columns
print("Overtime values:")
print(df["Overtime"].unique())

print("\nAttrition values:")
print(df["Attrition"].unique())

print("\nJob Satisfaction values:")
print(df["Job_Satisfaction"].unique())

print("\nWork-Life Balance values:")
print(df["Work_Life_Balance"].unique())

print("\nJob Level values:")
print(df["Job_Level"].unique())
```

```python
# Data Preparation Check

# Check for missing values
missing_values = df.isnull().sum()

# Check for duplicate rows
duplicate_rows = df.duplicated().sum()

print("Total Missing Values:", missing_values.sum())
print("Total Duplicate Rows:", duplicate_rows)

if missing_values.sum() == 0 and duplicate_rows == 0:
    print("\nDataset is clean and ready for EDA!")
else:
    print("\nData preparation is required.")
```

```python
# Feature Understanding

numerical_features = [
    "Age",
    "Monthly_Income",
    "Job_Satisfaction",
    "Years_at_Company",
    "Work_Life_Balance",
    "Job_Level"
]

print("Mean:")
display(df[numerical_features].mean())

print("\nMedian:")
display(df[numerical_features].median())

print("\nStandard Deviation:")
display(df[numerical_features].std())
```

```python
# Age Distribution

plt.figure(figsize=(8, 5))

sns.histplot(df["Age"], bins=10, kde=True)

plt.title("Age Distribution of Employees")
plt.xlabel("Age")
plt.ylabel("Number of Employees")
plt.tight_layout()

plt.show()
```

```python
# Monthly Income Distribution

plt.figure(figsize=(8, 5))

sns.histplot(df["Monthly_Income"], bins=10, kde=True)

plt.title("Monthly Income Distribution of Employees")
plt.xlabel("Monthly Income")
plt.ylabel("Number of Employees")
plt.tight_layout()

plt.show()
```

```python
# Job Satisfaction Distribution

plt.figure(figsize=(8, 5))

sns.countplot(x="Job_Satisfaction", data=df)

plt.title("Job Satisfaction Distribution")
plt.xlabel("Job Satisfaction Level")
plt.ylabel("Number of Employees")
plt.tight_layout()

plt.show()
```

```python
# Attrition Distribution

plt.figure(figsize=(8, 5))

sns.countplot(x="Attrition", data=df)

plt.title("Employee Attrition Distribution")
plt.xlabel("Attrition (0 = Stayed, 1 = Left)")
plt.ylabel("Number of Employees")
plt.xticks([0, 1], ["Stayed", "Left"])

plt.tight_layout()
plt.show()
```

```python
# Overtime vs Attrition

plt.figure(figsize=(8, 5))

sns.countplot(x="Overtime", hue="Attrition", data=df)

plt.title("Employee Attrition by Overtime")
plt.xlabel("Overtime")
plt.ylabel("Number of Employees")
plt.legend(title="Attrition", labels=["Stayed", "Left"])

plt.tight_layout()
plt.show()
```

```python
# Job Satisfaction vs Attrition

plt.figure(figsize=(8, 5))

sns.countplot(x="Job_Satisfaction", hue="Attrition", data=df)

plt.title("Employee Attrition by Job Satisfaction")
plt.xlabel("Job Satisfaction Level")
plt.ylabel("Number of Employees")
plt.legend(title="Attrition", labels=["Stayed", "Left"])

plt.tight_layout()
plt.show()
```

```python
# Years at Company vs Attrition

plt.figure(figsize=(8, 5))

sns.boxplot(x="Attrition", y="Years_at_Company", data=df)

plt.title("Years at Company vs Employee Attrition")
plt.xlabel("Attrition (0 = Stayed, 1 = Left)")
plt.ylabel("Years at Company")
plt.xticks([0, 1], ["Stayed", "Left"])

plt.tight_layout()
plt.show()
```

```python
# Monthly Income vs Attrition

plt.figure(figsize=(8, 5))

sns.boxplot(x="Attrition", y="Monthly_Income", data=df)

plt.title("Monthly Income vs Employee Attrition")
plt.xlabel("Attrition (0 = Stayed, 1 = Left)")
plt.ylabel("Monthly Income")
plt.xticks([0, 1], ["Stayed", "Left"])

plt.tight_layout()
plt.show()
```

```python
# Correlation Analysis

correlation_matrix = df.corr(numeric_only=True)

plt.figure(figsize=(10, 7))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5
)

plt.title("Correlation Heatmap of Employee Features")
plt.tight_layout()

plt.show()
```

```python
# Attrition Rate by Overtime

attrition_rate = df.groupby("Overtime")["Attrition"].mean() * 100

print("Attrition Rate by Overtime:")
print(attrition_rate)

plt.figure(figsize=(8, 5))

sns.barplot(
    x=attrition_rate.index,
    y=attrition_rate.values
)

plt.title("Attrition Rate by Overtime")
plt.xlabel("Overtime")
plt.ylabel("Attrition Rate (%)")

plt.tight_layout()
plt.show()
```

```python
# Key Questions and Findings

print("KEY QUESTIONS FROM THE DATA")
print("=" * 40)

# 1. Overall attrition rate
overall_attrition = df["Attrition"].mean() * 100
print(f"\n1. Overall Attrition Rate: {overall_attrition:.2f}%")

# 2. Average income by attrition
income_by_attrition = df.groupby("Attrition")["Monthly_Income"].mean()
print("\n2. Average Monthly Income:")
print(f"   Stayed: {income_by_attrition[0]:.2f}")
print(f"   Left:   {income_by_attrition[1]:.2f}")

# 3. Average years at company by attrition
years_by_attrition = df.groupby("Attrition")["Years_at_Company"].mean()
print("\n3. Average Years at Company:")
print(f"   Stayed: {years_by_attrition[0]:.2f}")
print(f"   Left:   {years_by_attrition[1]:.2f}")

# 4. Average job satisfaction by attrition
satisfaction_by_attrition = df.groupby("Attrition")["Job_Satisfaction"].mean()
print("\n4. Average Job Satisfaction:")
print(f"   Stayed: {satisfaction_by_attrition[0]:.2f}")
print(f"   Left:   {satisfaction_by_attrition[1]:.2f}")
```

```python
# Key Insights

overall_attrition = df["Attrition"].mean() * 100

overtime_rates = df.groupby("Overtime")["Attrition"].mean() * 100
satisfaction_rates = df.groupby("Job_Satisfaction")["Attrition"].mean() * 100

print("KEY INSIGHTS")
print("=" * 50)

print(f"\n1. Overall, {overall_attrition:.2f}% of employees in the dataset left the company.")

print(
    f"\n2. The attrition rate for employees with overtime was "
    f"{overtime_rates.get(1, 0):.2f}%."
)

print(
    f"\n3. The attrition rate for employees without overtime was "
    f"{overtime_rates.get(0, 0):.2f}%."
)

print("\n4. Attrition rates by Job Satisfaction level:")
for level, rate in satisfaction_rates.items():
    print(f"   Satisfaction Level {level}: {rate:.2f}%")

print("\n5. The correlation heatmap helps identify relationships")
print("   between employee characteristics and attrition.")

print("\n6. These findings can help identify employee groups")
print("   that may require further analysis or retention strategies.")
```

```python
# Final Conclusion

print("FINAL CONCLUSION")
print("=" * 50)

print("""
The Exploratory Data Analysis provided an overview of employee
characteristics and their relationship with attrition.

Statistical analysis was used to understand the distribution of
numerical features such as age, monthly income, job satisfaction,
years at the company, work-life balance, and job level.

Visualizations helped identify patterns in employee demographics,
income, satisfaction, overtime, and attrition.

Correlation analysis was also performed to understand relationships
between the numerical features and employee attrition.

Overall, the analysis demonstrates how Exploratory Data Analysis
can be used to understand employee data and identify patterns
that may be useful for further predictive modeling.
""")
```
